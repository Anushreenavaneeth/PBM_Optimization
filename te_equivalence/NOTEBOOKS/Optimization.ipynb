```python
import pandas as pd
import numpy as np
import joblib
import os

from IPython.display import display

print("Libraries imported.")
```

```python
candidate_path = (
    "G:\\PMS_Optimization\\NOTEBOOKS\\outputs\\PBM_Candidate_Alternatives.csv"
)

candidates = pd.read_csv(
    candidate_path
)

print(
    "Candidate dataset:",
    candidates.shape
)

display(
    candidates.head()
)
```

```python
model_path = (
    "G:\\PMS_Optimization\\models\\best_TE_model.pkl"
)

best_model = joblib.load(
    model_path
)

print(
    "Best TE model loaded successfully."
)
```

```python
features = [
    "Ingredient",
    "DF;Route",
    "Strength",
    "Appl_Type",
    "RLD",
    "RS",
    "Type",
    "Dosage_Form",
    "Route_Of_Administration",
    "Approved_Prior_To_1982"
]

X_candidates = candidates[
    features
].copy()

boolean_features = [
    "RLD",
    "RS",
    "Approved_Prior_To_1982"
]

for col in boolean_features:

    X_candidates[col] = (
        X_candidates[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "true": 1,
            "false": 0
        })
        .fillna(0)
    )

print(
    X_candidates.shape
)
```

```python
predicted_te = best_model.predict(
    X_candidates
)

candidates["Predicted_TE_Code"] = (
    predicted_te
)

print(
    "TE predictions generated."
)

display(
    candidates[
        [
            "Trade_Name",
            "TE_Code",
            "Predicted_TE_Code"
        ]
    ].head(30)
)
```

```python
print("Candidate columns:")
for i, col in enumerate(candidates.columns):
    print(i, ":", col)
```

```python
display(candidates.head())
```

```python
score_related_columns = [
    col for col in candidates.columns
    if any(word in col.lower()
           for word in ["patent", "exclusivity", "rs", "te"])
]

print("Scoring-related columns:")
print(score_related_columns)
```

```python
# ============================================================
# TE SCORE
# ============================================================

ab_family_codes = candidates["Predicted_TE_Code"].astype(str).str.upper().str.strip()

candidates["TE_Score"] = np.where(
    ab_family_codes.str.startswith("AB"),
    1.0,
    0.0
)

print("TE Score distribution:")
print(candidates["TE_Score"].value_counts())

display(
    candidates[
        [
            "Trade_Name",
            "TE_Code",
            "Predicted_TE_Code",
            "TE_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# PATENT SCORE
# ============================================================

candidates["Patent_Count"] = pd.to_numeric(
    candidates["Patent_Count"],
    errors="coerce"
).fillna(0)

max_patents = candidates["Patent_Count"].max()

if max_patents > 0:
    candidates["Patent_Score"] = (
        1 - candidates["Patent_Count"] / max_patents
    )
else:
    candidates["Patent_Score"] = 1.0

print("Patent Score created.")

display(
    candidates[
        [
            "Trade_Name",
            "Patent_Count",
            "Patent_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# EXCLUSIVITY SCORE
# ============================================================

candidates["Exclusivity_Count"] = pd.to_numeric(
    candidates["Exclusivity_Count"],
    errors="coerce"
).fillna(0)

max_exclusivity = candidates["Exclusivity_Count"].max()

if max_exclusivity > 0:
    candidates["Exclusivity_Score"] = (
        1 - candidates["Exclusivity_Count"] / max_exclusivity
    )
else:
    candidates["Exclusivity_Score"] = 1.0

print("Exclusivity Score created.")

display(
    candidates[
        [
            "Trade_Name",
            "Exclusivity_Count",
            "Exclusivity_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# RS SCORE
# ============================================================

candidates["RS_Score"] = (
    candidates["RS"]
    .fillna(False)
    .astype(str)
    .str.strip()
    .str.lower()
    .map({
        "true": 1.0,
        "false": 0.0,
        "1": 1.0,
        "0": 0.0
    })
    .fillna(0.0)
)

print("RS Score created.")

display(
    candidates[
        [
            "Trade_Name",
            "RS",
            "RS_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# FINAL PBM SCORE
# ============================================================

TE_WEIGHT = 0.40
PATENT_WEIGHT = 0.25
EXCLUSIVITY_WEIGHT = 0.20
RS_WEIGHT = 0.15

candidates["Final_Score"] = (
    TE_WEIGHT * candidates["TE_Score"]
    + PATENT_WEIGHT * candidates["Patent_Score"]
    + EXCLUSIVITY_WEIGHT * candidates["Exclusivity_Score"]
    + RS_WEIGHT * candidates["RS_Score"]
)

print("Final PBM score calculated.")

display(
    candidates[
        [
            "Trade_Name",
            "Predicted_TE_Code",
            "TE_Score",
            "Patent_Score",
            "Exclusivity_Score",
            "RS_Score",
            "Final_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# FIX PREDICTED TE CODE
# Convert numeric model prediction back to actual TE code
# ============================================================

te_label_map = {
    0: "AA",
    1: "AB",
    2: "AB1",
    3: "AB1,AB2",
    4: "AB1,AB2,AB3",
    5: "AB1,AB2,AB3,AB4",
    6: "AB1,AB3",
    7: "AB2",
    8: "AB3",
    9: "AB4",
    10: "AN",
    11: "AO",
    12: "AP",
    13: "AP1",
    14: "AP2",
    15: "AP3",
    16: "AT",
    17: "AT1",
    18: "AT2",
    19: "AT3",
    20: "BD",
    21: "BP",
    22: "BX"
}

candidates["Predicted_TE_Code"] = (
    pd.to_numeric(
        candidates["Predicted_TE_Code"],
        errors="coerce"
    )
    .map(te_label_map)
)

print("Converted predicted TE codes:")

display(
    candidates[
        [
            "Trade_Name",
            "TE_Code",
            "Predicted_TE_Code"
        ]
    ].head(30)
)
```

```python
# ============================================================
# RECALCULATE TE SCORE
# ============================================================

def calculate_te_score(te_code):

    if pd.isna(te_code):
        return 0.0

    te_code = str(te_code).strip().upper()

    # Best equivalence
    if te_code == "AB":
        return 1.0

    # Other acceptable AB categories
    elif te_code.startswith("AB"):
        return 0.8

    # Other TE categories
    elif te_code in ["AA", "AN", "AO", "AP", "AP1", "AP2", "AP3"]:
        return 0.5

    else:
        return 0.0


candidates["TE_Score"] = (
    candidates["Predicted_TE_Code"]
    .apply(calculate_te_score)
)

print("TE Score recalculated successfully.")

display(
    candidates[
        [
            "Trade_Name",
            "TE_Code",
            "Predicted_TE_Code",
            "TE_Score"
        ]
    ].head(30)
)
```

```python
# ============================================================
# RECALCULATE FINAL PBM SCORE
# ============================================================

candidates["Final_Score"] = (
    0.40 * candidates["TE_Score"]
    + 0.25 * candidates["Patent_Score"]
    + 0.20 * candidates["Exclusivity_Score"]
    + 0.15 * candidates["RS_Score"]
)

print("Final PBM Score recalculated.")

display(
    candidates[
        [
            "Trade_Name",
            "Predicted_TE_Code",
            "TE_Score",
            "Patent_Score",
            "Exclusivity_Score",
            "RS_Score",
            "Final_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# FINAL CANDIDATE RANKING
# ============================================================

ranked_candidates = (
    candidates
    .sort_values(
        by="Final_Score",
        ascending=False
    )
    .reset_index(drop=True)
)

ranked_candidates["Rank"] = (
    ranked_candidates.index + 1
)

print("Candidates ranked successfully.")

display(
    ranked_candidates[
        [
            "Rank",
            "Trade_Name",
            "Ingredient",
            "Strength",
            "Dosage_Form",
            "Route_Of_Administration",
            "Predicted_TE_Code",
            "TE_Score",
            "Patent_Score",
            "Exclusivity_Score",
            "RS_Score",
            "Final_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# BEST PBM ALTERNATIVE
# ============================================================

best_candidate = ranked_candidates.iloc[0]

print("=" * 60)
print("BEST PBM ALTERNATIVE")
print("=" * 60)

print("Trade Name       :", best_candidate["Trade_Name"])
print("Ingredient       :", best_candidate["Ingredient"])
print("Strength         :", best_candidate["Strength"])
print("Dosage Form      :", best_candidate["Dosage_Form"])
print("Route            :", best_candidate["Route_Of_Administration"])
print("Predicted TE     :", best_candidate["Predicted_TE_Code"])
print("TE Score         :", best_candidate["TE_Score"])
print("Patent Score     :", best_candidate["Patent_Score"])
print("Exclusivity Score:", best_candidate["Exclusivity_Score"])
print("RS Score         :", best_candidate["RS_Score"])
print("Final Score      :", best_candidate["Final_Score"])
print("Rank             :", best_candidate["Rank"])

print("=" * 60)
```
