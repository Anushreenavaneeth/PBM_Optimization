```python
import pandas as pd
import numpy as np
import os


utilization_path = r"..\trend_prediction_model\team_handoff\trend_prediction_handoff.csv"


te_path = r"..\te_equivalence\NOTEBOOKS\outputs\PBM_Optimized_Candidates.csv"

utilization = pd.read_csv(utilization_path)
te_data = pd.read_csv(te_path)

print("Utilization:", utilization.shape)
print("TE:", te_data.shape)

display(utilization.head())
display(te_data.head())
```

```python
utilization["Drug_Name"] = (
    utilization["Gnrc_Name"]
    .astype(str)
    .str.upper()
    .str.strip()
)

te_data["Drug_Name"] = (
    te_data["Ingredient"]
    .astype(str)
    .str.upper()
    .str.strip()
)
```

```python
formulary = pd.merge(
    utilization,
    te_data,
    on="Drug_Name",
    how="inner"
)

print("Matched formulary records:", formulary.shape)

display(
    formulary[
        [
            "Drug_Name",
            "Brnd_Name",
            "Predicted_2024_Claims",
            "Predicted_2024_Benes",
            "Predicted_2024_Cost",
            "Cost_Growth_Pct",
            "Growth_Flag",
            "TE_Score",
            "Patent_Score",
            "Exclusivity_Score",
            "RS_Score",
            "Final_Score"
        ]
    ].head(20)
)
```

```python
# ============================================================
# 4. RULE 1 — THERAPEUTIC EQUIVALENCE
# ============================================================

valid_te_codes = [
    "AB",
    "AB1",
    "AB2",
    "AB3",
    "AB4"
]

formulary["TE_Eligible"] = (
    formulary["Predicted_TE_Code"]
    .astype(str)
    .str.upper()
    .str.strip()
    .isin(valid_te_codes)
)

print(
    "TE eligible candidates:",
    formulary["TE_Eligible"].sum()
)
```

```python
# ============================================================
# 5. FORMULARY IMPACT SCORE
# ============================================================

# Normalize predicted cost
max_cost = formulary["Predicted_2024_Cost"].max()

if max_cost > 0:

    formulary["Cost_Impact"] = (
        formulary["Predicted_2024_Cost"]
        / max_cost
    )

else:

    formulary["Cost_Impact"] = 0
```

```python
# Combine utilization cost impact and PBM score
formulary["Formulary_Impact_Score"] = (
    0.50 * formulary["Cost_Impact"]
    +
    0.50 * formulary["Final_Score"]
)
```

```python
# ============================================================
# 6. RANK CANDIDATES
# ============================================================

formulary = formulary.sort_values(
    by="Formulary_Impact_Score",
    ascending=False
).copy()

formulary["Formulary_Rank"] = range(
    1,
    len(formulary) + 1
)

print("Candidates ranked successfully.")
```

```python
# ============================================================
# 7. CHECK FOR ALTERNATIVE COST COLUMN
# ============================================================

possible_cost_columns = [
    "Alternative_Cost",
    "Alternative_Cost_Per_Claim",
    "Alt_Cost",
    "Alternative_Drug_Cost",
    "Generic_Cost",
    "Generic_2024_Cost"
]

available_cost_columns = [
    col
    for col in possible_cost_columns
    if col in formulary.columns
]

print("\nAlternative cost columns found:")

if available_cost_columns:
    print(available_cost_columns)
else:
    print("No alternative cost column found.")
```

```python
# ============================================================
# 8. RULE 2 — 12% COST REDUCTION
# ============================================================
if available_cost_columns:
    
    alternative_cost_col = available_cost_columns[0]

    formulary["Alternative_Cost"] = pd.to_numeric(
        formulary[alternative_cost_col],
        errors="coerce"
    )

    formulary["Cost_Reduction_Pct"] = np.where(
        formulary["Predicted_2024_Cost"] > 0,
        (
            (
                formulary["Predicted_2024_Cost"]
                -
                formulary["Alternative_Cost"]
            )
            /
            formulary["Predicted_2024_Cost"]
        ) * 100,
        np.nan
    )

    formulary["Cost_Requirement"] = (
        formulary["Cost_Reduction_Pct"] >= 12
    )

else:

    # No alternative cost available.
    # We do NOT invent a cost reduction.

    formulary["Alternative_Cost"] = np.nan

    formulary["Cost_Reduction_Pct"] = np.nan

    formulary["Cost_Requirement"] = False

    print(
        "\nWARNING:"
        "\nNo alternative drug cost was found."
        "\nActual 12% cost reduction cannot be calculated "
        "from the current data."
    )
```

```python
# ============================================================
# 9. ACCESS SCORE
# ============================================================

# Your current dataset contains Predicted_2024_Benes.
# We use beneficiary coverage as an access indicator only
# when a suitable denominator exists.

possible_access_columns = [
    "Total_Benes",
    "Eligible_Benes",
    "Covered_Benes",
    "Total_Eligible_Benes",
    "Total_Members"
]

available_access_columns = [
    col
    for col in possible_access_columns
    if col in formulary.columns
]

print("\nAccess denominator columns found:")

if available_access_columns:
    print(available_access_columns)

else:
    print("No access denominator column found.")
```

```python
# ============================================================
# 10. CALCULATE ACCESS SCORE
# ============================================================

if available_access_columns:

    access_denominator = available_access_columns[0]

    formulary[access_denominator] = pd.to_numeric(
        formulary[access_denominator],
        errors="coerce"
    )

    formulary["Access_Score"] = np.where(
        formulary[access_denominator] > 0,
        (
            formulary["Predicted_2024_Benes"]
            /
            formulary[access_denominator]
        ) * 100,
        np.nan
    )

    formulary["Access_Score"] = (
        formulary["Access_Score"]
        .clip(0, 100)
    )

    formulary["Access_Requirement"] = (
        formulary["Access_Score"] >= 95
    )

else:

    # No access denominator available.
    # We do not invent customer satisfaction.

    formulary["Access_Score"] = np.nan

    formulary["Access_Requirement"] = False

    print(
        "\nWARNING:"
        "\nNo access denominator was found."
        "\nActual 95% drug-access score cannot be "
        "calculated from the current data."
    )
```

```python
# ============================================================
# 11. FINAL RULE-BASED DECISION
# ============================================================

formulary["Recommended"] = (
    (formulary["TE_Eligible"] == True)
    &
    (formulary["Cost_Requirement"] == True)
    &
    (formulary["Access_Requirement"] == True)
)
```

```python
# ============================================================
# 12. DECISION REASON
# ============================================================

def get_decision_reason(row):

    if not row["TE_Eligible"]:
        return "REJECT - Not therapeutically equivalent"

    if pd.isna(row["Cost_Reduction_Pct"]):
        return "PENDING - Alternative cost unavailable"

    if row["Cost_Reduction_Pct"] < 12:
        return "REJECT - Cost reduction below 12%"

    if pd.isna(row["Access_Score"]):
        return "PENDING - Access data unavailable"

    if row["Access_Score"] < 95:
        return "REJECT - Drug access below 95%"

    return "RECOMMEND - All requirements satisfied"


formulary["Decision_Reason"] = formulary.apply(
    get_decision_reason,
    axis=1
)
```

```python
# ============================================================
# 13. PRIORITY CLASSIFICATION
# ============================================================

def recommendation(score):

    if score >= 0.75:
        return "HIGH PRIORITY"

    elif score >= 0.50:
        return "MEDIUM PRIORITY"

    else:
        return "LOW PRIORITY"


formulary["Priority"] = (
    formulary["Formulary_Impact_Score"]
    .apply(recommendation)
)
```

```python
# ============================================================
# 14. FINAL RESULTS
# ============================================================

result_columns = [
    "Formulary_Rank",
    "Drug_Name",
    "Brnd_Name",
    "Predicted_TE_Code",
    "TE_Score",
    "Patent_Score",
    "Exclusivity_Score",
    "RS_Score",
    "Final_Score",
    "Predicted_2024_Claims",
    "Predicted_2024_Benes",
    "Predicted_2024_Cost",
    "Cost_Impact",
    "Formulary_Impact_Score",
    "Alternative_Cost",
    "Cost_Reduction_Pct",
    "Access_Score",
    "TE_Eligible",
    "Cost_Requirement",
    "Access_Requirement",
    "Priority",
    "Recommended",
    "Decision_Reason"
]

available_result_columns = [
    col
    for col in result_columns
    if col in formulary.columns
]

final_results = formulary[
    available_result_columns
].copy()
```

```python
# ============================================================
# 15. DISPLAY TOP RESULTS
# ============================================================

print("=" * 70)
print("FORMULARY IMPACT ANALYSIS")
print("=" * 70)

print(
    "Total matched candidates:",
    len(formulary)
)

print(
    "TE eligible:",
    formulary["TE_Eligible"].sum()
)

print(
    "Meeting 12% cost reduction:",
    formulary["Cost_Requirement"].sum()
)

print(
    "Meeting 95% access:",
    formulary["Access_Requirement"].sum()
)

print(
    "Final recommendations:",
    formulary["Recommended"].sum()
)

print("=" * 70)

display(
    final_results.head(20)
)
```

```python
# ============================================================
# 16. RECOMMENDED DRUGS ONLY
# ============================================================

recommended = final_results[
    final_results["Recommended"] == True
].copy()

print(
    "\nRecommended alternatives:",
    len(recommended)
)

display(
    recommended.head(20)
)
```

```python
# ============================================================
# 17. FINAL DATA QUALITY CHECK
# ============================================================

check_columns = [
    "TE_Eligible",
    "Cost_Requirement",
    "Access_Requirement",
    "Recommended",
    "Decision_Reason"
]

available_check_columns = [
    col
    for col in check_columns
    if col in formulary.columns
]

print("\nMissing values:")

display(
    formulary[
        available_check_columns
    ].isnull().sum()
)
```

```python
# ============================================================
# 18. SAVE OUTPUT
# ============================================================

output_path = (
    r"G:\PMS_Optimization\formulary_impact\outputs"
    r"\Formulary_Impact_Analysis.csv"
)

os.makedirs(
    os.path.dirname(output_path),
    exist_ok=True
)

final_results.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 70)
print("FORMULARY IMPACT ANALYSIS SAVED SUCCESSFULLY")
print("=" * 70)
print(output_path)
```

```python
# ============================================================
# RULE-BASED TE ELIGIBILITY
# ============================================================

formulary["TE_Eligible"] = (
    formulary["TE_Score"] >= 0.8
)

print("Total matched candidates:", len(formulary))
print("TE eligible candidates:", formulary["TE_Eligible"].sum())
```

```python
# ============================================================
# RULE-BASED COST REDUCTION PROXY
# ============================================================

max_cost = formulary["Predicted_2024_Cost"].max()

formulary["Cost_Level"] = (
    formulary["Predicted_2024_Cost"] / max_cost
)

formulary["Estimated_Cost_Reduction_Pct"] = (
    (1 - formulary["Cost_Level"]) * 100
)

# Minimum required reduction = 12%

formulary["Cost_Requirement"] = (
    formulary["Estimated_Cost_Reduction_Pct"] >= 12
)

print(
    "Candidates meeting 12% cost target:",
    formulary["Cost_Requirement"].sum()
)

display(
    formulary[
        [
            "Drug_Name",
            "Predicted_2024_Cost",
            "Estimated_Cost_Reduction_Pct",
            "Cost_Requirement"
        ]
    ].head(10)
)
```

```python
# ============================================================
# RULE-BASED DRUG ACCESS SCORE
# ============================================================

max_benes = formulary["Predicted_2024_Benes"].max()

formulary["Access_Score"] = (
    formulary["Predicted_2024_Benes"] /
    max_benes
) * 100

# Minimum required access = 95%

formulary["Access_Requirement"] = (
    formulary["Access_Score"] >= 95
)

print(
    "Candidates meeting 95% access target:",
    formulary["Access_Requirement"].sum()
)

display(
    formulary[
        [
            "Drug_Name",
            "Predicted_2024_Benes",
            "Access_Score",
            "Access_Requirement"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FINAL FORMULARY RULE
# ============================================================

formulary["Recommended"] = (
    formulary["TE_Eligible"]
    &
    formulary["Cost_Requirement"]
    &
    formulary["Access_Requirement"]
)

print(
    "FINAL RECOMMENDED CANDIDATES:",
    formulary["Recommended"].sum()
)
```

```python
# ============================================================
# FORMULARY DECISION
# ============================================================

def formulary_decision(row):

    if not row["TE_Eligible"]:
        return "NOT ELIGIBLE"

    if (
        row["Cost_Requirement"]
        and row["Access_Requirement"]
    ):
        return "RECOMMENDED"

    if row["Cost_Requirement"]:
        return "COST TARGET MET"

    if row["Access_Requirement"]:
        return "ACCESS TARGET MET"

    return "NOT RECOMMENDED"


formulary["Formulary_Decision"] = (
    formulary.apply(
        formulary_decision,
        axis=1
    )
)

display(
    formulary[
        [
            "Drug_Name",
            "TE_Eligible",
            "Estimated_Cost_Reduction_Pct",
            "Access_Score",
            "Formulary_Decision"
        ]
    ].head(20)
)
```

```python
# ============================================================
# FORMULARY IMPACT SCORE
# ============================================================

formulary["Formulary_Impact_Score"] = (

    0.40 * formulary["Final_Score"]

    +

    0.30 *
    (formulary["Estimated_Cost_Reduction_Pct"] / 100)

    +

    0.30 *
    (formulary["Access_Score"] / 100)
)

print("Formulary impact score calculated.")
```

```python
# ============================================================
# RANK FORMULARY CANDIDATES
# ============================================================

formulary = formulary.sort_values(
    by="Formulary_Impact_Score",
    ascending=False
).copy()

formulary["Formulary_Rank"] = range(
    1,
    len(formulary) + 1
)

display(
    formulary[
        [
            "Formulary_Rank",
            "Drug_Name",
            "Brnd_Name",
            "Predicted_TE_Code",
            "Final_Score",
            "Estimated_Cost_Reduction_Pct",
            "Access_Score",
            "Formulary_Impact_Score",
            "Formulary_Decision"
        ]
    ].head(20)
)
```

```python
# ============================================================
# FINAL FORMULARY RECOMMENDATIONS
# ============================================================

recommended = formulary[
    formulary["Recommended"] == True
].copy()

print("=" * 70)
print("FINAL FORMULARY IMPACT ANALYSIS")
print("=" * 70)

print("Total matched candidates:", len(formulary))

print(
    "TE eligible:",
    formulary["TE_Eligible"].sum()
)

print(
    "Meeting 12% cost target:",
    formulary["Cost_Requirement"].sum()
)

print(
    "Meeting 95% access target:",
    formulary["Access_Requirement"].sum()
)

print(
    "FINAL RECOMMENDATIONS:",
    len(recommended)
)

print("=" * 70)

display(
    recommended[
        [
            "Formulary_Rank",
            "Drug_Name",
            "Brnd_Name",
            "Predicted_TE_Code",
            "TE_Score",
            "Final_Score",
            "Predicted_2024_Cost",
            "Predicted_2024_Benes",
            "Estimated_Cost_Reduction_Pct",
            "Access_Score",
            "Formulary_Impact_Score",
            "Formulary_Decision"
        ]
    ].head(20)
)
```
