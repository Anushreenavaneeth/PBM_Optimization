```python
import pandas as pd

df = pd.read_csv(
    r"G:\PMS_Optimization\NOTEBOOKS\outputs\PBM_Optimized_Candidates.csv"
)

print("Candidates loaded:", len(df))
```

```python
drug_name = input("Enter drug name: ").strip().upper()

print("Searching for:", drug_name)
```

```python
matches = df[
    df["Ingredient"]
    .fillna("")
    .str.upper()
    .str.contains(drug_name, regex=False)
].copy()

print("Candidates found:", len(matches))
```

```python
recommendations = matches.sort_values(
    by="Final_Score",
    ascending=False
)
```

```python
display(
    recommendations[
        [
            "Rank",
            "Trade_Name",
            "Ingredient",
            "Strength",
            "Dosage_Form",
            "Route_Of_Administration",
            "Predicted_TE_Code",
            "TE_Score",
            "Patent_Score",
            "Exclusivity_Score",
            "RS_Score",
            "Final_Score"
        ]
    ].head(10)
)
```
